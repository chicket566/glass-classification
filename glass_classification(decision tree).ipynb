```python
X=df.iloc[:,0:9]
y=df.iloc[:,9]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV,train_test_split
from sklearn.metrics import accuracy_score,precision_score,f1_score,classification_report,confusion_matrix,recall_score
from sklearn.tree import DecisionTreeClassifier
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=0)
pipe=Pipeline([("model",DecisionTreeClassifier(random_state=42))
])
```

```python
Pipeline?
```

```python
param_grid = {
    "model__max_depth": [2, 3, 4, 5],
    "model__min_samples_leaf": [3, 5, 7, 10],
    "model__ccp_alpha": [0, 0.001, 0.01],
    "model__max_features": [None, "sqrt"]
}
scoring={"accuracy":"accuracy",
         "precision":"precision_macro",
          "recall":"recall_macro",
         "f1":"f1_macro"
}
grid=GridSearchCV(pipe,param_grid,cv=5,scoring=scoring,refit="accuracy",n_jobs=-1)
grid.fit(X_train,y_train)
```

```python
results=pd.DataFrame(grid.cv_results_)
pd.set_option({"display.max_rows":None,
    "display.max_columns":None
})
```

```python
#results[["params","mean_test_accuracy","mean_test_precision","mean_test_recall","mean_test_f1"]]
```

```python
y_pred=grid.predict(X_test)
y_pred
```

```python
y_prob=grid.predict_proba(X_test)[:,1]
y_prob
```

```python
accuracy_score(y_test,y_pred)
```

```python
precision_score(y_test,y_pred,average="macro")
```

```python
recall_score(y_test,y_pred,average="macro")
```

```python
f1_score(y_test,y_pred,average="macro")
```

```python
confusion_matrix(y_test,y_pred)
```

```python
report=pd.DataFrame(classification_report(y_test,y_pred,output_dict=True))
report
```

```python
accuracy_score(y_train,grid.predict(X_train))
```

```python
accuracy_score(y_test,grid.predict(X_test))
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20,10))

plot_tree(
    grid.best_estimator_["model"], 
    feature_names=X.columns,
    class_names=[str(c) for c in sorted(y.unique())],
    filled=True,
    rounded=True
)
plt.show()
```
